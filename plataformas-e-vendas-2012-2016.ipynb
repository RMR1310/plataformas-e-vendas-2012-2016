```python
# Primeira hipótese
# As classificações médias dos usuários das plataformas Xbox One e PC são as mesmas.
# Hipótese 1

xone_scores = df[df['platform'] == 'XOne']['user_score'].dropna()
pc_scores   = df[df['platform'] == 'PC']['user_score'].dropna()

print("Qtd de avaliações Xbox One:", len(xone_scores))
print("Qtd de avaliações PC:", len(pc_scores))
```

```python
t_stat, p_value = stats.ttest_ind(xone_scores, pc_scores, equal_var=False)

print("Estatística t:", t_stat)
print("p-valor:", p_value)
```

Conclusão da Hipótese 1
- Como p-valor < α, rejeitamos a hipótese nula (H₀).
- Isso significa que as médias de user_score para Xbox One e PC não são iguais.
- Há evidência estatística de que existe diferença significativa entre as avaliações dos usuários dessas duas plataformas.


```python
# Segunda hipótese
# As classificações médias dos usuários para os gêneros Action e Sports são diferentes.
# Hipótese 2

action_scores = df[df['genre'] == 'Action']['user_score'].dropna()
sports_scores = df[df['genre'] == 'Sports']['user_score'].dropna()

t_stat, p_value = stats.ttest_ind(action_scores, sports_scores, equal_var=False)

print("Estatística t:", t_stat)
print("p-valor:", p_value)

```

Conclusão da Hipótese 2
- Estatística t: 1.58 → indica uma diferença pequena entre as médias.
- p-valor: ≈ 0.115 → bem maior que o nível de significância α = 0.05.
 Como p-valor > α, não rejeitamos H₀.
Isso significa que não há evidência estatística suficiente para afirmar que as médias de user_score entre Action e Sports são diferentes.


Explicação
Na formulação das hipóteses, segui o padrão estatístico ensinado na Sprint 4:
- A hipótese nula (H₀) sempre representa a ideia de não diferença ou não efeito.
- Ex.: médias iguais entre Xbox One e PC; médias iguais entre Action e Sports.
- A hipótese alternativa (H₁) representa a possibilidade de diferença.
- Ex.: médias diferentes entre Xbox One e PC; médias diferentes entre Action e Sports.
Para o nível de significância (α) escolhi 0.05 (5%), porque é o valor mais usado em análises exploratórias. Ele equilibra rigor estatístico e praticidade:
- Se o p-valor ≤ 0.05, rejeitamos H₀ e concluímos que há diferença significativa.
- Se o p-valor > 0.05, não rejeitamos H₀ e consideramos que não há evidência suficiente de diferença
